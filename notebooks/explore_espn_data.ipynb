# ESPN Parsed Data Exploration

This notebook explores the parsed ESPN college basketball data stored in parquet format.

## Available Data
- **events** - Schedule events (games scheduled)
- **games** - Game metadata and results
- **teams** - Unique teams
- **venues** - Unique venues
- **players** - Unique players
- **boxscores** - Player box score statistics
- **playbyplay** - Play-by-play events

```python
import pandas as pd
from pathlib import Path

# Base path to parsed data
DATA_DIR = Path("../data/parsed/espn")

# Select league and season to explore
LEAGUE = "mens-college-basketball"
SEASON = 2025

# Build path to season data
season_dir = DATA_DIR / LEAGUE / str(SEASON)
print(f"Exploring: {season_dir}")
print(f"Available files: {[f.name for f in season_dir.glob('*.parquet')]}")
```

## Load All Data

```python
# Load all parquet files
events = pd.read_parquet(season_dir / "events.parquet")
games = pd.read_parquet(season_dir / "games.parquet")
teams = pd.read_parquet(season_dir / "teams.parquet")
venues = pd.read_parquet(season_dir / "venues.parquet")
players = pd.read_parquet(season_dir / "players.parquet")
boxscores = pd.read_parquet(season_dir / "boxscores.parquet")
playbyplay = pd.read_parquet(season_dir / "playbyplay.parquet")

# Summary of data sizes
print("Dataset Sizes:")
print(f"  Events:      {len(events):,} rows")
print(f"  Games:       {len(games):,} rows")
print(f"  Teams:       {len(teams):,} rows")
print(f"  Venues:      {len(venues):,} rows")
print(f"  Players:     {len(players):,} rows")
print(f"  BoxScores:   {len(boxscores):,} rows")
print(f"  PlayByPlay:  {len(playbyplay):,} rows")
```

## Box Scores

```python
print(f"BoxScores: {len(boxscores)} player-game records")
print(f"\nColumns: {list(boxscores.columns)}")
boxscores.head(10)
```

```python
# Top scorers (single game)
boxscores.nlargest(20, "PTS")[["athleteName", "teamName", "gid", "MIN", "PTS", "REB", "AST"]]
```

```python
# Season leaders (total points)
season_stats = boxscores.groupby(["aid", "athleteName", "teamName"]).agg({
    "gid": "count",
    "PTS": "sum",
    "REB": "sum",
    "AST": "sum",
    "MIN": "sum"
}).rename(columns={"gid": "games"}).reset_index()

# Points per game
season_stats["ppg"] = season_stats["PTS"] / season_stats["games"]
season_stats["rpg"] = season_stats["REB"] / season_stats["games"]
season_stats["apg"] = season_stats["AST"] / season_stats["games"]

# Filter to players with at least 10 games
qualified = season_stats[season_stats["games"] >= 10]
print(f"Players with 10+ games: {len(qualified)}")

# Top PPG
qualified.nlargest(20, "ppg")[["athleteName", "teamName", "games", "ppg", "rpg", "apg"]]
```

```python
# Shooting efficiency
boxscores["fg_pct"] = boxscores["FG_made"] / boxscores["FG_att"]
boxscores["three_pct"] = boxscores["3PT_made"] / boxscores["3PT_att"]
boxscores["ft_pct"] = boxscores["FT_made"] / boxscores["FT_att"]

# Team shooting averages
team_shooting = boxscores.groupby("teamName").agg({
    "FG_made": "sum",
    "FG_att": "sum",
    "3PT_made": "sum",
    "3PT_att": "sum",
    "FT_made": "sum",
    "FT_att": "sum"
}).reset_index()

team_shooting["fg_pct"] = team_shooting["FG_made"] / team_shooting["FG_att"]
team_shooting["three_pct"] = team_shooting["3PT_made"] / team_shooting["3PT_att"]
team_shooting["ft_pct"] = team_shooting["FT_made"] / team_shooting["FT_att"]

team_shooting.nlargest(15, "fg_pct")[["teamName", "fg_pct", "three_pct", "ft_pct"]]
```

## Cross-Dataset Analysis

```python
# Join games with team details
games_with_teams = games.merge(
    teams[["id", "conference"]].rename(columns={"id": "tm1_id", "conference": "tm1_conf"}),
    on="tm1_id",
    how="left"
).merge(
    teams[["id", "conference"]].rename(columns={"id": "tm2_id", "conference": "tm2_conf"}),
    on="tm2_id",
    how="left"
)

# Conference matchups
conf_matchups = games_with_teams.groupby(["tm1_conf", "tm2_conf"]).size().unstack(fill_value=0)
print("Conference matchup matrix (top conferences):")
top_confs = games_with_teams["tm1_conf"].value_counts().head(10).index
conf_matchups.loc[top_confs, top_confs]
```

```python
# Team performance summary
def get_team_record(team_id):
    """Calculate wins and losses for a team."""
    as_tm1 = completed[completed["tm1_id"] == team_id]
    as_tm2 = completed[completed["tm2_id"] == team_id]
    
    wins = len(as_tm1[as_tm1["tm1_winner"] == True]) + len(as_tm2[as_tm2["tm2_winner"] == True])
    losses = len(as_tm1[as_tm1["tm1_winner"] == False]) + len(as_tm2[as_tm2["tm2_winner"] == False])
    
    return wins, losses

# Build standings
standings = []
for _, team in teams.iterrows():
    wins, losses = get_team_record(team["id"])
    if wins + losses > 0:
        standings.append({
            "team": team["displayName"],
            "conference": team["conference"],
            "wins": wins,
            "losses": losses,
            "win_pct": wins / (wins + losses)
        })

standings_df = pd.DataFrame(standings).sort_values("win_pct", ascending=False)
standings_df.head(25)
```
